### Get Tokens From Dataset

```python
def get_tokens(dataset):
    gt_token_sequences_list = []
    added_tokens_list = []
    
    for sample in dataset:
        ground_truth = sample["ground_truth"]
        gt_token_sequences, added_tokens = process_ground_truth(tokenizer, ground_truth)
        
        gt_token_sequences_list.append(gt_token_sequences)
        added_tokens_list.append(added_tokens)
        
    return gt_token_sequences_list, added_tokens_list
```
